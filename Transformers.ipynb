```python
from transformers import pipeline
```

```python
tg = pipeline(task="text-generation", model="Qwen/Qwen2.5-1.5B")
```

```python
text = "what is transformers model in AI and examplain how it's used"
```

```python
res=tg(text, max_length=100, max_new_tokens=200, do_sample=True, temperature=0.9)
print(res)
```
